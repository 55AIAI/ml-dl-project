## 1. Theoretical Summary of Six Classification Algorithms

### 1.1 k-Nearest Neighbors (k-NN)

**Core Mathematical Idea:**  
k-NN is a non-parametric, instance-based (lazy) learner. Classification is based on the majority vote of the `k` closest training examples in the feature space. Given a query point `x`, distances to all training points are computed (commonly Euclidean: `d(x,xᵢ) = √Σ(xⱼ - xᵢⱼ)²`), and the `k` nearest neighbors vote on the class label.

**Key Hyperparameters:**
- `k` (number of neighbors): Small k → high variance (overfitting); Large k → high bias (underfitting)
- Distance metric: Euclidean, Manhattan, Minkowski
- Weighting: Uniform vs. distance-weighted votes

**Complexity:**
- Training: O(1) — no explicit training phase
- Inference: O(n·d) where n = number of training samples, d = dimensions
- Space: O(n·d) — stores entire training set

**Best Use Cases:** Small datasets, low-dimensional data, recommendation systems, anomaly detection.  
**Limitations:** Slow inference on large datasets, suffers from curse of dimensionality, sensitive to irrelevant features and scale, requires feature normalization.

---

### 1.2 Decision Trees

**Core Mathematical Idea:**  
Decision Trees recursively partition the feature space by selecting the feature and threshold that maximizes a purity criterion. For classification, the most common criteria are:
- **Gini Impurity:** `Gini = 1 - Σpᵢ²`
- **Information Gain (Entropy):** `H = -Σ pᵢ log₂(pᵢ)`

At each node, the split that maximizes Information Gain (or minimizes Gini) is chosen. Leaves represent class predictions.

**Key Hyperparameters:**
- `max_depth`: Controls tree depth; deeper trees overfit
- `min_samples_split` / `min_samples_leaf`: Regularize tree growth
- `criterion`: gini vs entropy

**Complexity:**
- Training: O(n·d·log(n)) — for balanced trees
- Inference: O(log(n)) per sample
- Space: O(n) in worst case (fully grown tree)

**Best Use Cases:** Interpretability-required settings, rule extraction, mixed data types.  
**Limitations:** Prone to overfitting, unstable (small data changes → different trees), biased toward features with more levels.

---

### 1.3 Random Forest

**Core Mathematical Idea:**  
Random Forest is an ensemble method combining multiple Decision Trees via bootstrap aggregating (bagging). Each tree is trained on a random bootstrap sample of the data, and at each split only a random subset of features (`√d` for classification) is considered. Final prediction is the majority vote across all trees:
```
ŷ = mode{ h₁(x), h₂(x), ..., hₜ(x) }
```
This decorrelates the trees, reducing variance without increasing bias.

**Key Hyperparameters:**
- `n_estimators`: Number of trees; more trees = better but slower
- `max_features`: Subset of features per split
- `max_depth`, `min_samples_split`: Per-tree regularization

**Complexity:**
- Training: O(T·n·d·log(n)) where T = number of trees
- Inference: O(T·log(n))
- Space: O(T·n)

**Best Use Cases:** Tabular data, feature importance analysis, high-dimensional classification.  
**Limitations:** Less interpretable than single trees, slow for real-time inference with many trees, memory-intensive.

---

### 1.4 Support Vector Machines (SVM)

**Core Mathematical Idea:**  
SVM seeks the maximum-margin hyperplane that separates classes. For linearly separable data:
```
Minimize: (1/2)||w||²
Subject to: yᵢ(w·xᵢ + b) ≥ 1 for all i
```
The margin is `2/||w||`. For non-linear problems, the **kernel trick** maps data to higher dimensions: RBF kernel: `K(x,z) = exp(-γ||x-z||²)`.

**Key Hyperparameters:**
- `C`: Regularization — small C = soft margin (more misclassification allowed)
- `kernel`: linear, rbf, poly
- `gamma`: RBF bandwidth; controls decision boundary curvature

**Complexity:**
- Training: O(n²) to O(n³) depending on solver
- Inference: O(n_sv · d) where n_sv = number of support vectors

**Best Use Cases:** High-dimensional data, text classification, small-to-medium datasets.  
**Limitations:** Slow on large datasets, difficult to interpret, requires careful kernel/hyperparameter selection, sensitive to feature scaling.

---

### 1.5 Logistic Regression

**Core Mathematical Idea:**  
Logistic Regression models the posterior probability of class membership using the logistic (sigmoid) function:
```
P(y=1|x) = σ(w·x + b) = 1 / (1 + e^(-(w·x+b)))
```
Parameters are learned by maximizing the log-likelihood (equivalent to minimizing cross-entropy loss) via gradient descent or Newton's method.

**Key Hyperparameters:**
- `C` (inverse of regularization strength)
- `penalty`: L1 (Lasso, sparsity), L2 (Ridge, weight decay)
- `solver`: lbfgs, saga, liblinear

**Complexity:**
- Training: O(n·d·k) where k = iterations
- Inference: O(d) per sample

**Best Use Cases:** Binary/multiclass classification, probability calibration, baseline model, interpretable linear boundaries.  
**Limitations:** Assumes linear decision boundary, poor on complex non-linear problems, requires feature engineering.

---

### 1.6 Naive Bayes

**Core Mathematical Idea:**  
Naive Bayes applies Bayes' theorem with the **naive** conditional independence assumption between features:
```
P(y|x) ∝ P(y) · Π P(xⱼ|y)
```
For Gaussian NB: `P(xⱼ|y) = N(xⱼ; μⱼ,y, σ²ⱼ,y)` — each feature modeled as Gaussian per class.

**Key Hyperparameters:**
- `var_smoothing` (GaussianNB): Prevents zero probabilities
- Distribution type: Gaussian, Multinomial, Bernoulli

**Complexity:**
- Training: O(n·d) — extremely fast
- Inference: O(d·C) where C = number of classes

**Best Use Cases:** Text classification (spam), real-time prediction, high-dimensional data, baseline classifier.  
**Limitations:** Strong independence assumption rarely holds in practice, poor probability calibration, sensitive to correlated features.

## 3. Algorithm Deep Dive — Random Forest

### 3.1 Step-by-Step Mathematical Derivation

**Step 1: Bootstrap Sampling**  
Given training set `D = {(x₁,y₁), ..., (xₙ,yₙ)}`, for each tree `t = 1,...,T`:  
- Draw bootstrap sample `Dₜ` of size n **with replacement** from D
- ~63.2% unique samples per bootstrap (probability of inclusion = 1-(1-1/n)ⁿ → 1-1/e)

**Step 2: Feature Subsampling at Each Node**  
At each split node, randomly select `m = √d` features (for classification).  
Only these `m` features are considered for the best split.

**Step 3: Gini Impurity Calculation**  
For a node with class distribution `{p₁, p₂, ..., pC}`:  
```
Gini(node) = 1 - Σᶜ pᶜ²
```
For a candidate split into left (L) and right (R) children:  
```
Gini_split = (nL/n)·Gini(L) + (nR/n)·Gini(R)
```
Choose the feature+threshold minimizing `Gini_split`.

**Step 4: Grow Tree Until Stopping Criterion**  
Repeat Step 3 recursively until: max_depth reached, min_samples_split not met, or node is pure.

**Step 5: Aggregation (Bagging)**  
```
ŷ = argmax_c Σᵀₜ₌₁ 𝟙[hₜ(x) = c]
```
The class with the most votes among all T trees is the final prediction.

**Why it Works — Bias-Variance Decomposition:**  
```
Error = Bias² + Variance + Noise
```
- Each tree has low bias (deep, complex)
- Bootstrap sampling + feature randomness decorrelates trees
- Averaging T decorrelated estimators: `Var(avg) ≈ ρσ² + (1-ρ)σ²/T`  
  where ρ = pairwise correlation. As ρ → 0, variance → 0 as T → ∞

### 3.2 Hand-Crafted Example

**Dataset (4 samples, 2 features, binary class):**

| x₁ | x₂ | y |
|----|----|---|
| 1  | 2  | 0 |
| 2  | 1  | 0 |
| 4  | 5  | 1 |
| 5  | 4  | 1 |

**Bootstrap sample (T=1):** Draw {sample1, sample1, sample3, sample4}  
**Selected feature (m=√2≈1):** Feature x₁

**Candidate split x₁ < 3:**
- Left: {x₁=1,y=0}, {x₁=1,y=0} → p(0)=1, Gini(L) = 1-1² = 0
- Right: {x₁=4,y=1}, {x₁=5,y=1} → p(1)=1, Gini(R) = 0
- Gini_split = (2/4)·0 + (2/4)·0 = **0.0** (perfect split!)

Tree prediction for new point x=[3.5, 3]: x₁=3.5 ≥ 3 → **class 1** ✓

### 3.3 Overfitting and Underfitting

**Overfitting (high variance):**
- Very deep trees with small min_samples_leaf → each tree memorizes training data
- Low n_estimators (e.g., T=1 is just a single decision tree)
- Detection: Training accuracy >> Validation accuracy
- Fix: Increase min_samples_leaf, reduce max_depth, increase T

**Underfitting (high bias):**
- max_depth=1 → shallow stumps can't capture complex patterns
- Very small max_features (m=1) → insufficient information at each split
- Detection: Both training and validation accuracy are low
- Fix: Increase max_depth, increase max_features, use more estimators

````python
# ============================================================
#  Task 1 — Deep Dive Visualization: Random Forest
#  Decision Boundary Illustration on 2D synthetic data
# ============================================================
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from sklearn.datasets import make_classification
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier

np.random.seed(42)

# Generate 2D dataset for visualization
X, y = make_classification(
    n_samples=300, n_features=2, n_redundant=0,
    n_informative=2, n_clusters_per_class=1, random_state=42
)

# Create meshgrid for decision boundary
x_min, x_max = X[:, 0].min() - 0.5, X[:, 0].max() + 0.5
y_min, y_max = X[:, 1].min() - 0.5, X[:, 1].max() + 0.5
xx, yy = np.meshgrid(np.linspace(x_min, x_max, 300),
                     np.linspace(y_min, y_max, 300))

fig, axes = plt.subplots(1, 3, figsize=(18, 5))
fig.suptitle("Random Forest — Decision Boundaries vs Single Tree vs Deep Tree",
             fontsize=14, fontweight='bold')

configs = [
    (DecisionTreeClassifier(max_depth=1, random_state=42), "Decision Stump\n(max_depth=1, Underfitting)"),
    (DecisionTreeClassifier(max_depth=None, random_state=42), "Deep Decision Tree\n(max_depth=None, Overfitting)"),
    (RandomForestClassifier(n_estimators=100, random_state=42), "Random Forest\n(n_estimators=100)"),
]

colors = ['#FF6B6B', '#4ECDC4']
cmap_bg = plt.cm.RdYlGn

for ax, (clf, title) in zip(axes, configs):
    clf.fit(X, y)
    Z = clf.predict_proba(np.c_[xx.ravel(), yy.ravel()])[:, 1]
    Z = Z.reshape(xx.shape)
    
    ax.contourf(xx, yy, Z, levels=50, cmap=cmap_bg, alpha=0.6)
    ax.contour(xx, yy, Z, levels=[0.5], colors='black', linewidths=2)
    
    scatter = ax.scatter(X[:, 0], X[:, 1], c=y, cmap=plt.cm.bwr,
                         edgecolors='k', s=40, alpha=0.8)
    
    train_acc = clf.score(X, y)
    ax.set_title(f"{title}\nTrain Acc: {train_acc:.3f}", fontsize=11)
    ax.set_xlabel("Feature 1")
    ax.set_ylabel("Feature 2")

p0 = mpatches.Patch(color='blue', label='Class 0')
p1 = mpatches.Patch(color='red', label='Class 1')
fig.legend(handles=[p0, p1], loc='lower center', ncol=2, fontsize=10)
plt.tight_layout()
plt.savefig('task1_decision_boundaries.png', dpi=150, bbox_inches='tight')
plt.show()
print("Figure saved: task1_decision_boundaries.png")
````

````python
# ============================================================
#  Task 1 — Gini Impurity Visualization
# ============================================================
import numpy as np
import matplotlib.pyplot as plt

p = np.linspace(0, 1, 200)
gini = 2 * p * (1 - p)          # Gini for binary class
entropy = -np.where(p > 0, p * np.log2(np.clip(p, 1e-9, 1)), 0) \
          -np.where((1-p) > 0, (1-p) * np.log2(np.clip(1-p, 1e-9, 1)), 0)
misclass = 1 - np.maximum(p, 1 - p)

fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(p, gini,     label='Gini Impurity',      color='#E74C3C', linewidth=2.5)
ax.plot(p, entropy,  label='Entropy',color='#3498DB', linewidth=2.5,
        linestyle='--')
ax.plot(p, 0.5 * entropy, label='0.5 × Entropy', color='#3498DB', linewidth=1, linestyle=':')
ax.plot(p, misclass, label='Misclass. Error',     color='#2ECC71', linewidth=2.5,
        linestyle='-.')

ax.set_xlabel("Fraction of class 1 (p)", fontsize=12)
ax.set_ylabel("Impurity Value", fontsize=12)
ax.set_title("Impurity Measures for Binary Classification", fontsize=13, fontweight='bold')
ax.legend(fontsize=11)
ax.axvline(0.5, color='gray', linestyle=':', alpha=0.7)
ax.set_xlim(0, 1)
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('task1_impurity_measures.png', dpi=150, bbox_inches='tight')
plt.show()
print("Figure saved: task1_impurity_measures.png")
````
